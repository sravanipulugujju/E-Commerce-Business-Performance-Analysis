# E-Commerce Data CleaningStep-by-step cleaning process: load data -> explore -> fix data types -> handle missing values column by column -> final check.

## 1. Import Libraries

```python
import pandas as pd
```

## 2. Load Dataset

```python
ecommerce = pd.read_csv("/content/ecommerce.csv")
ecommerce
```

## 3. Initial Data Exploration

```python
ecommerce.info()
```

```python
ecommerce.dtypes
```

```python
ecommerce.isnull().sum()
```

## 4. Clean Order_Date Check missing values, convert to proper datetime format, then fill missing dates with the median date.

```python
ecommerce["Order_Date"].isnull().sum()
```

```python
ecommerce[ecommerce["Order_Date"].isnull()]
```

```python
ecommerce["Order_Date"] = pd.to_datetime(
    ecommerce["Order_Date"],
    format='%d/%m/%Y'
)
ecommerce["Order_Date"]
```

```python
ecommerce["Order_Date"].median()
```

```python
ecommerce["Order_Date"] = ecommerce["Order_Date"].fillna(ecommerce["Order_Date"].median())
ecommerce["Order_Date"]
```

```python
ecommerce["Order_Date"].isnull().sum()
```

## 5. Clean Category Check missing values and fill with the most frequent value (mode).

```python
ecommerce["Category"].isnull().sum()
```

```python
ecommerce[ecommerce["Category"].isnull()]
```

```python
ecommerce["Category"].mode()
```

```python
ecommerce["Category"] = ecommerce["Category"].fillna(ecommerce["Category"].mode()[0])
ecommerce["Category"]
```

```python
ecommerce["Category"].isnull().sum()
```
